## **BUSINESS UNDERSTANDING**

Karamoja, the most food-insecure region of Uganda, faces significant challenges in agricultural productivity, due to intense droughts, pest infestations, and disease outbreaks. These factors have led to low crop yields, particularly for staple crops such as sorghum and maize.

##**DATA CLEANING**

Data cleaning is an important aspect in analsis as through this we would be able to find errors in the data, fix them and this would ensure data accuracy and reliability during matchine learning.

```python
#checking for duplcates
[df1.duplicated().sum(), df2.duplicated().sum()]
```

```python
#checking for missing values
[df1.isnull().sum(), df2.isnull().sum()]
```

It is also noted that there are no missing values in the individual datasets as shown in the above code so we will head straight to combining the datasets and filling the missing values since we noted that df1 and df2 have different data shape.

```python
# combining the data frames by columns
combined_df = pd.merge(df1, df2, how='outer')
df.head()
```

The reason for using outer join is to enable us to get all records from both the tables. From the warning provided above that i am merging int and float values, I will solve it in the following codes.

```python
# filling the missing values in the column region with mean for numeric value
numeric_columns = combined_df.select_dtypes(include=[np.number]).columns
combined_df[numeric_columns] = combined_df[numeric_columns].fillna(combined_df[numeric_columns].mean())
```

For the non-numeric column instead of filling the missing values with a place holder i will opt for the mode this will enable us to have the missing values filled with the most repeated words.

```python
# filling missing values for non-numeric columns
non_numeric_columns = combined_df.select_dtypes(exclude=[np.number]).columns
combined_df[non_numeric_columns] = combined_df[non_numeric_columns].fillna(combined_df[non_numeric_columns].mode().iloc[0])
```

Now that we have managed to fill the missing values and combine them, we can now download the combined file as **"Karamoja_combined_df.csv"**

```python
# saving the combined dataset
combined_df.to_csv('Karamoja_combined_df.csv', index=False)
```
